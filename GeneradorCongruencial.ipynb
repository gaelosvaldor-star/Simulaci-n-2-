```python
def generador(a, c, m, x0):
  x1 = (a * x0 + c) % m
  return x1
```

```python
a = 7
b = 5
m = 8
x0 = 1
n = 10
muestra = []
datos = []
for i in range(n):
  datos.append(x0) # Collect current x0
  nuevo = generador(a, b, m, x0) # Call the corrected function
  muestra.append([nuevo])
  x0 = nuevo
datos.append(x0) # Append the last x0 after the loop finishes
print(muestra)
```
