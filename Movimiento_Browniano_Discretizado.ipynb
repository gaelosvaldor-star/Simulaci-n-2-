```python
import random as r
import numpy as np
import matplotlib.pyplot as plt
```

```python
def brown(T, N):
    lista = []
    dt = T / N
    w_0 = 0
    for i in range(1, N):
        dw = r.gauss(0, 1)
        w_1 = w_0 + np.sqrt(dt) * dw
        lista.append(w_1)
        w_0 = w_1
    return lista
```

```python
M = brown(1,1000)
```

```python
plt.plot(M)
```
