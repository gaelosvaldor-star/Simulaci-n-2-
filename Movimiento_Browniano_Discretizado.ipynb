```python
import random as r
import numpy as np
import matplotlib.pyplot as plt
```

```python
def brown(T, N):
    lista = [0.0]
    dt = T / N
    w_0 = 0.0
    for i in range(1, N):
        dw = r.gauss(0, 1)
        w_1 = w_0 + np.sqrt(dt) * dw
        lista.append(w_1)
        w_0 = w_1
    return lista
```

```python
def trayectoria(n):
  for i in range(n):
    M = brown(1, 1000)
    plt.plot(M)
  plt.show()
```

```python
trayectoria(1000)
```

```python
def refi(lista, T):
  n = len(lista)
  dt = np.sqrt(T / n)
  nueva_lista = [lista[0]]

  for i in range(n - 1):
    dw = r.gauss(0, 1)
    w_medio = 0.5 * (lista[i] + lista[i+1]) + 0.5 * dw * dt
    nueva_lista.append(w_medio)
    nueva_lista.append(lista[i+1])

  return nueva_lista
```

```python
L = brown(1,5)
```

```python
R=refi(L,1)
```

```python
plt.plot(R)
plt.plot(L)
```
